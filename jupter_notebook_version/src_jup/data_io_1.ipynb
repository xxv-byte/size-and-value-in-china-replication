```python
# 资产负债表数据
def find_bs_data():
    files = list(rawdata_dir.rglob(bs_pattern))
    files = sorted(files)
    return files

def read_bs_data():
    files = find_bs_data()
    data_list = []
    for file in files:
        data = pd.read_csv(file ,
                           usecols = [
                               bs_stock_code , 
                               bs_end_date , 
                               bs_public_date , 
                               total_ass , 
                               total_cur_assets , 
                               total_cur_lia , 
                               short_loan , 
                               long_loan , 
                               tax_pay , 
                               cash_eqv , 
                               trad_ass , 
                               total_liab
                           ] , 
                           dtype = {
                               bs_stock_code : 'string' , 
                               total_ass : 'float' , 
                               total_cur_assets : 'float' , 
                               total_cur_lia : 'float' , 
                               short_loan : 'float' , 
                               long_loan : 'float' , 
                               tax_pay : 'float' , 
                               cash_eqv : 'float' , 
                               trad_ass : 'float' , 
                               total_liab : 'float'
                           } , 
                           parse_dates = [
                               bs_end_date , 
                               bs_public_date
                           ] , 
                           low_memory = False
                           )
        data = data[data[bs_end_date] <= final_date]
        data_list.append(data)
    df = pd.concat(data_list , ignore_index = True)
    return df
```

```python
# 现金流量表数据
def find_CF_data():
    files = list(rawdata_dir.rglob(cf_pattern))
    files = sorted(files)
    return files

def read_cf_data():
    """读取构造 CP 与应计利润必需的现金流字段；股票代码按字符串读取以保留前导零。"""
    files = find_CF_data()

    data_list = []
    for file in files:
        data_list.append(
            pd.read_csv(
                file,
                usecols=[cf_code ,
                         cf_report_type , 
                         cf_end_date , 
                         cf_public_date , 
                         cf_net_increase , 
                         cf_intangible_amort , 
                         cf_defer_amort],
                dtype={
                    cf_code: 'string',
                    cf_report_type: 'string',
                    cf_net_increase: 'float',
                    cf_intangible_amort: 'float',
                    cf_defer_amort: 'float'
                },
                parse_dates=[cf_end_date , 
                             cf_public_date],
                low_memory=False
            )
        )
    cf_df = pd.concat(data_list , ignore_index = True)
    return cf_df
```
