```python
# 规定data_io.ipynb的读取路径
import os

# 项目根目录：优先环境变量 RESSET_PROJECT_ROOT，否则自当前目录向上查找含 data/raw 的目录。
# 发布副本中不含任何个人绝对路径。
def _find_project_root(start=None):
    env_root = os.environ.get('RESSET_PROJECT_ROOT')
    if env_root:
        return Path(env_root).expanduser().resolve()
    start = Path.cwd() if start is None else Path(start).resolve()
    for candidate in [start] + list(start.parents):
        if (candidate / 'data' / 'raw').exists():
            return candidate
    raise RuntimeError(
        '未找到项目根目录：请把本仓库放在含 data/raw 的目录下，'
        '或设置环境变量 RESSET_PROJECT_ROOT。'
    )


root = _find_project_root()
# 数据根目录：默认 <root>/data；若要把样例/异地数据指给流水线，设置 RESSET_DATA_DIR。
data_dir = Path(
    os.environ.get('RESSET_DATA_DIR', root / 'data')
).expanduser().resolve()
rawdata_dir = data_dir / 'raw'

daily_stock_pattern = 'RESSET_DRESSTK_*.csv'
lsdata_pattern = 'RESSET_THIMKTSTKINFO_1.csv'
month_pattern = 'RESSET_MRESSTK_*.csv'
merger_pattern = 'RESSET_MERG_1.csv'
cf_pattern = 'RESSET_SCF_ALL_1.csv'
bs_pattern = 'RESSET_BS_ALL_1.csv'
fas_pattern = 'RESSET_BSNOTE_FIXASSDEPR_*.csv'
```
